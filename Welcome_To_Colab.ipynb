```
# Usually pre-installed in Colab, but just in case
!pip install seaborn scikit-learn xgboost
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
```

```
from google.colab import files
uploaded = files.upload()
df = pd.read_csv("Worldwide Travel Cities Dataset (Ratings and Climate).csv")
```

```
# Drop unnecessary columns
df_cleaned = df.drop(columns=["id", "short_description", "latitude", "longitude", "avg_temp_monthly", "ideal_durations", "city", "country"])

# Encode target variable (classification: budget_level)
df_cleaned["budget_level_encoded"] = df_cleaned["budget_level"].map({
    "Budget": 0, "Mid-range": 1, "Luxury": 2
})
df_cleaned = df_cleaned.drop(columns=["budget_level"])
```

```
# One-hot encode categorical column
region_encoded = pd.get_dummies(df_cleaned["region"], prefix="region")

# Combine with numerical features
features = pd.concat([region_encoded, df_cleaned.drop(columns=["region", "budget_level_encoded"])], axis=1)
target = df_cleaned["budget_level_encoded"]

# Combine for EDA
eda_df = pd.concat([features, target], axis=1)
```

```
X_train, X_test, y_train, y_test = train_test_split(
    features, target, test_size=0.2, random_state=42, stratify=target
)
```

```
sns.set(style="whitegrid")

eda_df.drop(columns="budget_level_encoded").hist(
    bins=20, figsize=(16, 12), color="skyblue", edgecolor="black"
)

plt.suptitle("Feature Distributions", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()
```

```
plt.figure(figsize=(14, 10))
sns.heatmap(eda_df.corr(), annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

```
# List of all numerical features
features_to_plot = ['culture', 'adventure', 'nature', 'beaches', 'nightlife',
                    'cuisine', 'wellness', 'urban', 'seclusion']

# Create boxplots for each feature by budget level
plt.figure(figsize=(16, 20))
for i, feature in enumerate(features_to_plot, 1):
    plt.subplot(5, 2, i)
    sns.boxplot(x=target, y=df_cleaned[feature])
    plt.xlabel("Budget Level (Encoded)")
    plt.ylabel(feature.capitalize())
    plt.title(f"{feature.capitalize()} by Budget Level")

plt.tight_layout()
plt.show()
```

```
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# Initialize models
logreg = LogisticRegression(max_iter=1000, random_state=42)
dtree = DecisionTreeClassifier(random_state=42)
rf = RandomForestClassifier(random_state=42)

# Fit models on training data
logreg.fit(X_train, y_train)
dtree.fit(X_train, y_train)
rf.fit(X_train, y_train)

# Predict on test data
y_pred_logreg = logreg.predict(X_test)
y_pred_dtree = dtree.predict(X_test)
y_pred_rf = rf.predict(X_test)

# Predict on train data for later comparison
y_train_logreg = logreg.predict(X_train)
y_train_dtree = dtree.predict(X_train)
y_train_rf = rf.predict(X_train)
```

```
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

# Store model names and predictions
models = {
    "Logistic Regression": (y_train_logreg, y_pred_logreg),
    "Decision Tree": (y_train_dtree, y_pred_dtree),
    "Random Forest": (y_train_rf, y_pred_rf)
}

# Evaluation loop
for name, (y_train_pred, y_test_pred) in models.items():
    print(f"\n📌 {name}")
    print("→ Train Accuracy:", accuracy_score(y_train, y_train_pred))
    print("→ Test Accuracy :", accuracy_score(y_test, y_test_pred))
    print("→ F1 Score (Test):", f1_score(y_test, y_test_pred, average='weighted'))
    print("→ Classification Report (Test):\n", classification_report(y_test, y_test_pred))
    print("→ Confusion Matrix (Test):\n", confusion_matrix(y_test, y_test_pred))
```

```
# Collect results into lists
model_names = ["Logistic Regression", "Decision Tree", "Random Forest"]
test_accuracies = [
    accuracy_score(y_test, y_pred_logreg),
    accuracy_score(y_test, y_pred_dtree),
    accuracy_score(y_test, y_pred_rf)
]
test_f1_scores = [
    f1_score(y_test, y_pred_logreg, average='weighted'),
    f1_score(y_test, y_pred_dtree, average='weighted'),
    f1_score(y_test, y_pred_rf, average='weighted')
]

# Plot
plt.figure(figsize=(10, 5))
bar_width = 0.35
x = np.arange(len(model_names))

plt.bar(x - bar_width/2, test_accuracies, width=bar_width, label='Accuracy', color='skyblue')
plt.bar(x + bar_width/2, test_f1_scores, width=bar_width, label='F1 Score', color='salmon')

plt.xlabel('Model')
plt.ylabel('Score')
plt.title('Model Performance Comparison (Test Set)')
plt.xticks(x, model_names)
plt.ylim(0, 1)
plt.legend()
plt.grid(True, axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report, roc_auc_score
from sklearn.preprocessing import label_binarize
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Prepare for ROC-AUC: Binarize target for multi-class
y_test_binarized = label_binarize(y_test, classes=[0, 1, 2])

# Predict probabilities for ROC-AUC
y_score_logreg = logreg.predict_proba(X_test)
y_score_dtree = dtree.predict_proba(X_test)
y_score_rf = rf.predict_proba(X_test)

# Store results
model_names = ["Logistic Regression", "Decision Tree", "Random Forest"]
train_accuracies = []
test_accuracies = []
train_f1_scores = []
test_f1_scores = []
roc_aucs = []

# Loop through models
models = {
    "Logistic Regression": (logreg, y_train_logreg, y_pred_logreg, y_score_logreg),
    "Decision Tree": (dtree, y_train_dtree, y_pred_dtree, y_score_dtree),
    "Random Forest": (rf, y_train_rf, y_pred_rf, y_score_rf)
}

for name, (model, y_train_pred, y_test_pred, y_proba) in models.items():
    print(f"\n📌 {name}")

    # Classification report
    print("→ Classification Report (Test):\n", classification_report(y_test, y_test_pred))

    # Confusion matrix
    print("→ Confusion Matrix (Test):\n", confusion_matrix(y_test, y_test_pred))

    # Accuracy & F1
    train_accuracies.append(accuracy_score(y_train, y_train_pred))
    test_accuracies.append(accuracy_score(y_test, y_test_pred))
    train_f1_scores.append(f1_score(y_train, y_train_pred, average='weighted'))
    test_f1_scores.append(f1_score(y_test, y_test_pred, average='weighted'))

    # ROC-AUC (macro averaged)
    roc_auc = roc_auc_score(y_test_binarized, y_proba, multi_class="ovr", average="macro")
    roc_aucs.append(roc_auc)

# Create summary table
summary_df = pd.DataFrame({
    "Model": model_names,
    "Train Accuracy": train_accuracies,
    "Test Accuracy": test_accuracies,
    "Train F1 Score": train_f1_scores,
    "Test F1 Score": test_f1_scores,
    "ROC-AUC (Macro)": roc_aucs
}).round(3)

print("\n✅ Summary Table:")
display(summary_df)
```

```
# 📦 Step 1: Import libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# 📁 Step 2: Load dataset
df = pd.read_csv("/content/Worldwide Travel Cities Dataset (Ratings and Climate) (1).csv")

# 🔤 Step 3: Encode target variable
label_encoder = LabelEncoder()
df['budget_level_encoded'] = label_encoder.fit_transform(df['budget_level'])  # Budget=0, Luxury=1, Mid=2

# 🎯 Step 4: Select features and target
features = ['culture', 'adventure', 'nature', 'beaches', 'nightlife',
            'cuisine', 'wellness', 'urban', 'seclusion']
X = df[features]
y = df['budget_level_encoded']

# ⚖️ Step 5: Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 🔀 Step 6: Train-test split (80:20)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# 🔍 Step 7: Check missing values
print("Missing values:\n", df.isnull().sum())

# 📊 Step 8: EDA - Histograms
df[features].hist(bins=20, figsize=(15, 10), edgecolor='black')
plt.suptitle('Histograms of Numerical Features', fontsize=16)
plt.tight_layout()
plt.show()

# 📦 EDA - Boxplots
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[features])
plt.title('Boxplot of Features')
plt.xticks(rotation=90)
plt.show()

# 📉 EDA - Countplot for target
sns.countplot(data=df, x='budget_level')
plt.title('Budget Level Distribution')
plt.show()

# 📈 EDA - Correlation heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(df[features].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()

# 🧪 Step 9: Evaluation function
def evaluate_model(name, model, X_train, X_test, y_train, y_test):
    print(f"\n====== {name} ======")

    # Predictions
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    # Accuracy
    train_acc = accuracy_score(y_train, y_train_pred)
    test_acc = accuracy_score(y_test, y_test_pred)

    print(f"Train Accuracy: {round(train_acc, 3)}")
    print(f"Test Accuracy: {round(test_acc, 3)}")

    # Classification report
    print("\nClassification Report (Test):")
    print(classification_report(y_test, y_test_pred))

    # Confusion matrix
    cm = confusion_matrix(y_test, y_test_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.title(f'{name} - Confusion Matrix (Test Set)')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()

    # Store results
    return {
        'Model': name,
        'Train Accuracy': train_acc,
        'Test Accuracy': test_acc
    }

# 🤖 Step 10: Train and evaluate Logistic Regression
lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train, y_train)
lr_result = evaluate_model("Logistic Regression", lr_model, X_train, X_test, y_train, y_test)

# 🌳 Step 11: Train and evaluate Decision Tree
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)
dt_result = evaluate_model("Decision Tree", dt_model, X_train, X_test, y_train, y_test)

# 🌲 Step 12: Train and evaluate Random Forest
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)
rf_result = evaluate_model("Random Forest", rf_model, X_train, X_test, y_train, y_test)

# 🧾 Step 13: Final comparison table
summary = pd.DataFrame([lr_result, dt_result, rf_result])
print("\n🔍 Final Model Comparison:")
print(summary.sort_values(by='Test Accuracy', ascending=False).reset_index(drop=True))
```
